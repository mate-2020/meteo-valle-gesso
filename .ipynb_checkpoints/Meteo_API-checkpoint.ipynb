```python
#!/usr/bin/env python3
"""
Multi-Model Weather Comparison - Valli del Cuneese
Modelli: ECMWF IFS, ICON-D2, AROME France HD
Località: Entracque, Valdieri, Sant'Anna di Valdieri,
          Terme di Valdieri, Castelmagno, Acceglio, Bellino
Fonte: Open-Meteo API (gratuita, senza chiave)
Versione: 6.0 - Pulizia pre-esecuzione + stampa su schermo
"""

import requests
import json
import glob
import os
from datetime import datetime
from typing import Dict, Optional

# ============================================================
#  CONFIGURAZIONE
# ============================================================

LOCATIONS = {
    "Entracque":            {"lat": 44.250, "lon": 7.400},
    "Valdieri":             {"lat": 44.283, "lon": 7.400},
    "Sant'Anna di Valdieri":{"lat": 44.233, "lon": 7.317},
    "Terme di Valdieri":    {"lat": 44.205, "lon": 7.271},
    "Castelmagno":          {"lat": 44.417, "lon": 7.217},
    "Acceglio":             {"lat": 44.475, "lon": 6.989},
    "Bellino":              {"lat": 44.583, "lon": 7.033},
}

MODELS = {
    "ECMWF":   "ecmwf_ifs025",
    "ICON-D2": "icon_d2",
    "AROME":   "meteofrance_arome_france_hd",
}

BASE_URL = "https://api.open-meteo.com/v1/forecast"
CURRENT_VARS = "temperature_2m,wind_speed_10m,precipitation"
HOURLY_VARS  = "temperature_2m,precipitation,rain,wind_speed_10m"
DAILY_VARS   = "temperature_2m_max,temperature_2m_min,precipitation_sum,rain_sum"

# ============================================================
#  UTILITY
# ============================================================

def cleanup_old_files():
    """Cancella tutti i vecchi file meteo_cuneo_*.txt e *.json nella cartella corrente"""
    deleted_txt = 0
    deleted_json = 0
    
    for pattern in ["meteo_cuneo_*.txt", "meteo_cuneo_*.json"]:
        for filepath in glob.glob(pattern):
            try:
                os.remove(filepath)
                print(f"  Eliminato: {filepath}")
                if filepath.endswith(".txt"):
                    deleted_txt += 1
                else:
                    deleted_json += 1
            except OSError as e:
                print(f"  Errore eliminando {filepath}: {e}")
    
    total_deleted = deleted_txt + deleted_json
    if total_deleted == 0:
        print("  Nessun file vecchio da eliminare")
    else:
        print(f"  Totale eliminati: {deleted_txt} TXT, {deleted_json} JSON")
    
    return total_deleted

def safe_float(value, default=0.0):
    if value is None:
        return default
    try:
        return float(value)
    except (TypeError, ValueError):
        return default

def safe_round(value, decimals=1):
    f = safe_float(value, None)
    if f is None:
        return None
    return round(f, decimals)

def calculate_spread(values):
    """Calcola mean, spread, min, max da una lista di valori"""
    clean = [v for v in values if v is not None]
    if len(clean) < 2:
        return {
            "mean": safe_round(clean[0]) if clean else None,
            "spread": 0,
            "min": safe_round(clean[0]) if clean else None,
            "max": safe_round(clean[0]) if clean else None,
        }
    return {
        "mean": safe_round(sum(clean) / len(clean)),
        "spread": safe_round(max(clean) - min(clean)),
        "min": safe_round(min(clean)),
        "max": safe_round(max(clean)),
    }

# ============================================================
#  FETCH DETERMINISTIC
# ============================================================

def fetch_model(model_label: str, model_id: str, lat: float, lon: float) -> Optional[Dict]:
    params = {
        "latitude": lat,
        "longitude": lon,
        "current": CURRENT_VARS,
        "hourly": HOURLY_VARS,
        "daily": DAILY_VARS,
        "timezone": "Europe/Rome",
        "forecast_days": 3,
        "models": model_id,
    }
    try:
        resp = requests.get(BASE_URL, params=params, timeout=15)
        resp.raise_for_status()
        return resp.json()
    except requests.exceptions.HTTPError:
        print(f"HTTP {resp.status_code}", end="")
        return None
    except Exception as e:
        print(f"{type(e).__name__}", end="")
        return None

def extract_metrics(data) -> Optional[Dict]:
    if not data:
        return None

    current = data.get("current", {})
    daily = data.get("daily", {})
    hourly = data.get("hourly", {})

    now_temp = safe_round(current.get("temperature_2m"))
    now_wind = safe_round(current.get("wind_speed_10m"))
    now_precip = safe_round(current.get("precipitation"))

    t_max = daily.get("temperature_2m_max", [])
    t_min = daily.get("temperature_2m_min", [])
    precip_sum = daily.get("precipitation_sum", [])
    rain_sum = daily.get("rain_sum", [])

    today_max = safe_round(t_max[0]) if len(t_max) > 0 else None
    today_min = safe_round(t_min[0]) if len(t_min) > 0 else None
    today_precip = safe_round(precip_sum[0]) if len(precip_sum) > 0 else None
    tomorrow_precip = safe_round(precip_sum[1]) if len(precip_sum) > 1 else None

    times = hourly.get("time", [])
    rains = hourly.get("rain", [])
    rain_24h = 0.0
    if times and rains:
        now_hour = datetime.now().strftime("%Y-%m-%dT%H")
        start_idx = 0
        for i, t in enumerate(times):
            if str(t).startswith(now_hour):
                start_idx = i
                break
        rain_24h = safe_round(sum(safe_float(r) for r in rains[start_idx:start_idx + 24]))

    return {
        "now_temp": now_temp,
        "now_wind": now_wind,
        "now_precip": now_precip,
        "today_max": today_max,
        "today_min": today_min,
        "today_precip": today_precip,
        "tomorrow_precip": tomorrow_precip,
        "rain_24h": rain_24h,
    }

# ============================================================
#  SPREAD SIMULATO
# ============================================================

def compute_model_spread(det_results: Dict) -> Dict:
    temps_today = []
    precips_today = []
    precips_tomorrow = []

    for label, m in det_results.items():
        if m:
            if m.get("today_max") is not None:
                temps_today.append(m["today_max"])
            if m.get("today_precip") is not None:
                precips_today.append(m["today_precip"])
            if m.get("tomorrow_precip") is not None:
                precips_tomorrow.append(m["tomorrow_precip"])

    temp_spread = calculate_spread(temps_today)
    precip_spread = calculate_spread(precips_today)
    precip_tomorrow_spread = calculate_spread(precips_tomorrow)

    ts = temp_spread["spread"] or 0
    ps = precip_spread["spread"] or 0

    if ts < 3 and ps < 5:
        confidence = "ALTA"
        marker = "[ + ]"
    elif ts < 6 and ps < 15:
        confidence = "MEDIA"
        marker = "[ ~ ]"
    else:
        confidence = "BASSA"
        marker = "[ ! ]"

    return {
        "temp_spread": temp_spread,
        "precip_spread": precip_spread,
        "precip_tomorrow_spread": precip_tomorrow_spread,
        "confidence": confidence,
        "marker": marker,
    }

# ============================================================
#  REPORT
# ============================================================

def fmt(val, suffix="", width=8):
    if val is None:
        return f"{'--':>{width}}"
    s = f"{val}{suffix}"
    return f"{s:>{width}}"

def generate_report(loc_name, coords, det_results, spread):
    L = []
    L.append("")
    L.append("=" * 78)
    L.append(f"  {loc_name.upper()} — {coords['lat']}°N, {coords['lon']}°E")
    L.append("=" * 78)
    L.append("")

    L.append("  MODELLI DETERMINISTICI")
    L.append("-" * 78)
    L.append(f"  {'Modello':<10} {'T ora':>7} {'Vento':>7} {'Pioggia':>8} {'Precip':>8} {'Precip':>8} {'T max':>7} {'T min':>7}")
    L.append(f"  {'':<10} {'(°C)':>7} {'(km/h)':>7} {'ora(mm)':>8} {'oggi(mm)':>8} {'domani':>8} {'oggi':>7} {'oggi':>7}")
    L.append("-" * 78)

    for label in MODELS:
        m = det_results.get(label)
        if not m:
            L.append(f"  {label:<10}  {'--':>6}  {'--':>6}  {'--':>7}  {'--':>7}  {'--':>7}  {'--':>6}  {'--':>6}")
            continue
        L.append(
            f"  {label:<10} "
            f"{fmt(m.get('now_temp'), '°C'):>7} "
            f"{fmt(m.get('now_wind')):>7} "
            f"{fmt(m.get('now_precip')):>8} "
            f"{fmt(m.get('today_precip')):>8} "
            f"{fmt(m.get('tomorrow_precip')):>8} "
            f"{fmt(m.get('today_max'), '°C'):>7} "
            f"{fmt(m.get('today_min'), '°C'):>7}"
        )
    L.append("-" * 78)
    L.append("")

    ts = spread["temp_spread"]
    ps = spread["precip_spread"]
    pts = spread["precip_tomorrow_spread"]

    L.append("  DIVERGENZA TRA MODELLI (SPREAD SIMULATO)")
    L.append("-" * 78)
    L.append(f"  Temperatura max oggi:")
    L.append(f"    Media:  {fmt(ts.get('mean'), '°C')}")
    L.append(f"    Min:    {fmt(ts.get('min'), '°C')}")
    L.append(f"    Max:    {fmt(ts.get('max'), '°C')}")
    L.append(f"    Spread: {fmt(ts.get('spread'), '°C')}")
    L.append("")
    L.append(f"  Precipitazione oggi:")
    L.append(f"    Media:  {fmt(ps.get('mean'), 'mm')}")
    L.append(f"    Min:    {fmt(ps.get('min'), 'mm')}")
    L.append(f"    Max:    {fmt(ps.get('max'), 'mm')}")
    L.append(f"    Spread: {fmt(ps.get('spread'), 'mm')}")
    L.append("")
    L.append(f"  Precipitazione domani:")
    L.append(f"    Media:  {fmt(pts.get('mean'), 'mm')}")
    L.append(f"    Spread: {fmt(pts.get('spread'), 'mm')}")
    L.append("")

    L.append(f"  {spread['marker']} AFFIDABILITA': {spread['confidence']}")
    if spread["confidence"] == "BASSA":
        L.append(f"      Modelli discordanti — previsione incerta")
    elif spread["confidence"] == "MEDIA":
        L.append(f"      Moderata divergenza tra modelli")
    else:
        L.append(f"      Modelli concordanti")

    L.append("")
    L.append("=" * 78)
    return "\n".join(L)

def generate_summary(all_results):
    L = []
    L.append("")
    L.append("=" * 78)
    L.append("  RIASSUNTO COMPARATIVO")
    L.append("=" * 78)
    L.append("")
    L.append(f"  {'Località':<25} {'T ora':>7} {'Precip oggi':>11} {'T max':>7} {'Spread T':>9} {'Conf.':>7}")
    L.append("-" * 78)

    for loc_name, res in all_results.items():
        det = res["deterministic"]
        spread = res["spread"]

        first = next(iter(det.values())) if det else None
        t_now = first.get("now_temp") if first else None
        p_today = first.get("today_precip") if first else None
        t_max = spread["temp_spread"].get("mean")
        t_spread = spread["temp_spread"].get("spread")
        conf = spread["confidence"]

        L.append(
            f"  {loc_name:<25} "
            f"{fmt(t_now, '°C'):>7} "
            f"{fmt(p_today, 'mm'):>11} "
            f"{fmt(t_max, '°C'):>7} "
            f"{fmt(t_spread, '°C'):>9} "
            f"{conf:>7}"
        )
    L.append("-" * 78)
    L.append("")
    return "\n".join(L)

# ============================================================
#  MAIN
# ============================================================

def main():
    now_str = datetime.now().strftime("%Y%m%d_%H%M%S")
    report_dt = datetime.now().strftime("%d/%m/%Y %H:%M")

    print("=" * 78)
    print("  METEO MULTIMODELLO - VALLI DEL CUNEESE")
    print(f"  Esecuzione: {report_dt}")
    print("=" * 78)
    print()

    # STEP 1: Cancella vecchi file
    print("  STEP 1: Pulizia cartella...")
    cleaned = cleanup_old_files()
    print()

    print("  STEP 2: Scaricamento dati...")
    print("-" * 78)

    all_results = {}

    for loc_name, coords in LOCATIONS.items():
        print(f"\n  [{loc_name}] {coords['lat']}°N {coords['lon']}°E")
        det_results = {}
        for label, model_id in MODELS.items():
            print(f"    -> {label}...", end=" ")
            data = fetch_model(label, model_id, coords["lat"], coords["lon"])
            if data:
                metrics = extract_metrics(data)
                det_results[label] = metrics
                if metrics:
                    print(f"OK (T: {metrics.get('now_temp')}°C)")
                else:
                    print("PARZIALE")
            else:
                det_results[label] = None
                print("FAIL")

        spread = compute_model_spread(det_results)
        all_results[loc_name] = {
            "deterministic": det_results,
            "spread": spread,
        }

    print("\n" + "-" * 78)
    print("  STEP 3: Generazione report...")
    print()

    # Genera tutto il report
    full_report = []
    full_report.append("")
    full_report.append("=" * 78)
    full_report.append("  REPORT METEO MULTIMODELLO")
    full_report.append(f"  Valli della Provincia di Cuneo")
    full_report.append(f"  Generato: {report_dt}")
    full_report.append("=" * 78)

    for loc_name in LOCATIONS:
        res = all_results[loc_name]
        full_report.append(generate_report(
            loc_name, LOCATIONS[loc_name], res["deterministic"], res["spread"]
        ))

    full_report.append(generate_summary(all_results))

    full_report.append("")
    full_report.append("=" * 78)
    full_report.append("  LEGENDA")
    full_report.append("=" * 78)
    full_report.append("")
    full_report.append("  Modelli:")
    full_report.append("    ECMWF   = ECMWF IFS 0.25° (~25 km, globale, 15 giorni)")
    full_report.append("    ICON-D2 = DWD ICON-D2 (~2 km, Europa centrale, 2 giorni)")
    full_report.append("    AROME   = Meteo-France AROME HD (~1.5 km, Francia, 4 giorni)")
    full_report.append("")
    full_report.append("  Spread = divergenza tra i 3 modelli (non ensemble a 51 membri)")
    full_report.append("    Spread T < 3°C e P < 5mm  -> ALTA affidabilità")
    full_report.append("    Spread T < 6°C e P < 15mm -> MEDIA affidabilità")
    full_report.append("    Altrimenti                -> BASSA affidabilità")
    full_report.append("")
    full_report.append("  Per decisioni critiche: www.arpa.piemonte.it")
    full_report.append("")

    report_string = "\n".join(full_report)

    # Stampa su schermo
    print(report_string)

    # Salva su file
    txt_file = f"meteo_cuneo_{now_str}.txt"
    with open(txt_file, "w", encoding="utf-8") as f:
        f.write(report_string)

    json_data = {
        "timestamp": datetime.now().isoformat(),
        "locations": {
            loc: {
                "coordinates": LOCATIONS[loc],
                "deterministic": {
                    lbl: m for lbl, m in all_results[loc]["deterministic"].items() if m
                },
                "spread": all_results[loc]["spread"],
            }
            for loc in LOCATIONS
        },
    }
    json_file = f"meteo_cuneo_{now_str}.json"
    with open(json_file, "w", encoding="utf-8") as f:
        json.dump(json_data, f, indent=2, ensure_ascii=False, default=str)

    print("\n" + "-" * 78)
    print("  OUTPUT SALVATI:")
    print(f"    {txt_file}")
    print(f"    {json_file}")
    print("\n" + "=" * 78)
    print("  COMPLETATO")
    print("=" * 78)

if __name__ == "__main__":
    try:
        main()
    except KeyboardInterrupt:
        print("\n\n  Interrotto.")
    except Exception as e:
        import traceback
        print(f"\n  ERRORE: {type(e).__name__}: {e}")
        traceback.print_exc()
```
